```python
#PY PROGRAMME TO FIND NO.OF DUPLICATES IN A LIST
s=list(map(int,input().split()))
p=set(s)
duplicates=len(s)-len(p)
print(duplicates)
```

```python
#PY PROGRAMME TO FIND HOW MANY NUMBERS ARE HAVING DUPLICATES
L=[int(i) for i in input().split()]
L2=[]
count=0
for i in L:
  if L.count(i)>1:
    L2.append(i)
print(L2)
s=set(L2)
print(s)
print(len(s))
```

```python
#PROG TO PRINT NON DUPLICATE NUMBERS
L=[int(i) for i in input().split()]
L2=[]
count=0
for i in L:
  if L.count(i)==1:
    L2.append(i)
print(L2)
```

```python
#WRITE A PY PROG THAT PRINTS MISSING NUMBERS BETWEEN 1 TO N
N=int(input())
L=list(map(int,input().split()))
L2=[]
s=set(L)
for i in range(1,N+1):
  if i not in s:
    L2.append(i)
    print(L2)
```
