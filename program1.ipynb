```python
#ARTHEMATIC OPERATORS
a=int(input())
b=int(input())
print("sum of two numbers is",a+b)
print("difference of two numbers is",a-b)
print("product of two numbers is",a*b)
print("division of two numbers is",a/b)
print("modulo operator:",a%b)
print("floor division:",a//b)
print("exponentiation:" ,a**b)
```

```python
#COMPARISON OPERATORS
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

```python
#LOGICAL OPERATORS
print(a<b and a>b)
print(a>b or a>b)
print(not(a>b and b<a))
```

```python
#IDENTITY OPERATORS
print(a is b)
print(a is not b)
```

```python
#ASSIGNMEBT OPERATORS
print(a=b)
print(a+=b)
print(a-=b)
print(a*=b)
print(a/=b)
print(a//=b)
print(a**=b)
print(a%=b)
print(a>>=b)
print(a<<=b)
print()
```
