```python
#Program that finds max of three numbers
a=int(input("Enter: "))
b=int(input("Enter: "))
c=int(input("Enter: "))
if(a>b and a>c):
  print(f"{a} is greatest")
elif(b>a and b>c):
  print(f"{b} is greatest")
else:
  print(f"{c} is greatest")
```

```python
#Given year is leap year or not
a=int(input())
if((a%4==0 and a%100!=0) or (a%400==0)):
  print("LEAP YEAR")
else:
  print("NOT A LEAP YEAR")
```

```python
#MATCH CASE
day=int(input())
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3 :
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("invalid")
```

```python
#PROGRAM TO PRINT REVERSE OF A NUMBER
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum*10+rem
  n=n//10
print(sum)
```

```python
#PROGRAM THAT CHECKS GIVEN NUMBER IS PRIME OR NOT
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("Not a prime")
    break
else:
  print("Prime")
```

```python
#PROGRAM THAT PRINTS PRIME NUMBERS BETWEEN TWO given NUMBERS
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=" ")


```
