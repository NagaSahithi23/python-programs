```python
#PROGRAM ON "W" MODE
file=open("f1.txt","w")
data="hello everyone"
file.write(data)
print("file created")
file.close()
```

```python
#PROGRAM ON "r" MODE
with open("f1.txt","r") as f:
  data=f.read()
print(data)
```

```python
#PROGRAM ON "a" MODE
with open("f1.txt","a") as f:
  f.writelines(["\nname","\nage","\nplace"])
print("done")
```

```python
#READ LINE
with open("f1.txt","r") as f:
  print(f.readline())
```

```python
#READ LINES
with open("f1.txt","r") as f:
  print(f.readlines())
```

```python
#COPYING CONTENT OF ONE FILE INTO ANOTHER FILE
with open("f1.txt","r") as f:
  data=f.read()
print("reading done")
with open("f2.txt","w") as f2:
  f2.write(data)
print("content copied")
```

```python
#PY PROGRAM THAT PRINTS EACH LINE OF A LINE IN REVERSE ORDER
with open("f1.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```

```python
with open("f1.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  data_sort=sorted(data_lower)
with open("f4.txt","w") as f2:
  for i in data_sort:
    f2.write(i+"\n")
  print("done")
```

```python
#PROGRAM THAT COMPUTE THE NUMBER OF CHARACTERS,WORDS AND LINES IN A FILE
file=open("new.txt","w")
data="This is Python Programming !"
file.write(data)
file.close()
with open("new.txt","r") as f:
  content=f.read()
char=len(content)
words=len(content.split())
lines=len(content.splitlines())
print("no.of characters:",char)
print("no.of words:",words)
print("no.of lines:",lines)
```

```python
import pickle
data={1:2,3:4}
file=open("data.bin","wb")
pickle.dump(data,file)
file.close()
file=open("data.bin","rb")
pickle.load(file)
```
